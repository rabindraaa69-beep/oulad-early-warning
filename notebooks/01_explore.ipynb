```python
import pandas as pd
from pathlib import Path

RAW = Path("../data/raw")

tables = {}
for name in ["courses", "assessments", "vle", "studentInfo",
             "studentRegistration", "studentAssessment"]:
    tables[name] = pd.read_csv(RAW / f"{name}.csv")
    print(f"{name:22s} rows={tables[name].shape[0]:>8,}  cols={tables[name].shape[1]}")
```

```python
info = tables["studentInfo"]
print(info.columns.tolist())
print(info["final_result"].value_counts())
print(info["final_result"].value_counts(normalize=True).round(3))
info.head()
```

```python
clicks = pd.read_csv(
    RAW / "studentVle.csv",
    dtype={"code_module": "category", "code_presentation": "category",
           "id_student": "int32", "id_site": "int32",
           "date": "int16", "sum_click": "int16"},
)
print(clicks.shape)
print(round(clicks.memory_usage(deep=True).sum() / 1e6), "MB in memory")
clicks.head()
```
